```python
username="admin"
password="1234"
max_att=3 # maximum attempts
att=0 # intial taken no.of attemps
while att<max_att: # here the loop runs until it reachs to maximum no.of attempts 
    a=input("Enter Username: ") # taking username from user
    b=input("Enter Password: ") # taking password from user
    if a==username and b==password: # checking the username & password is correct ot not
        print("Login Successful")
        break # if both username & password is correct then loop has been exit
    else:
        att+= 1 # increases attempts if username and password is incorrect
        print(f'Invalid Credentials Attempts left: {max_att-att}') # showing remain attempts
else:
    print("Account Locked")
```

```python
marks=[45, 78, 90, 33, 60] # list of student results
pass_count=0 # pass count intilizing
fail_count=0 # fail count intilizing
min_marks=50 # minimum pass marks
for m in marks: #Analyzing each student marks
    if m>=min_marks:
        pass_count+=1 # passed student count
    else:
        fail_count+=1 # failed student count
print("Total Students :", len(marks))
print("no.of passed   :", pass_count)
print("no.of failed   :", fail_count)
```

```python
names=[" Alice ", "bob", " CHARLIE "] # raw data
clean_data=[] # here we are taking an empty list for storing an cleaned data
for n in names: # checking each name using loop
    data=n.strip() # here we are removing the spaces
    data=data.lower() # here we are converting the data into lowercase
    clean_data.append(data) # adding cleaned data to empty list
print("Cleaned List:",clean_data)
```

```python
messages=["Hi", "Welcome to the platform", "OK"]
for m in messages: # checking the each message using loop
    print(m, "Length:", len(m)) # printing the message and the message length
    if len(m) > 10: # here we are checking the message length is greater than 10 or not
        print("Too Long\n") # if the message is greater than 10 it prints Too Long
    else:
        print("OK") # if the message is lessthan 10 it prints OK
```

```python
logs=["INFO", "ERROR", "WARNING", "ERROR"]
count=0 # here we are taking count for errors
for l in logs: # here we are using the loop for checking each log entry
    if l=="ERROR": # Checking if the log is consist error
        count+=1   # here it will increase's the count if the condition is true
print('logs analysis\n')
print("Total ERROR entries:",count)
```
