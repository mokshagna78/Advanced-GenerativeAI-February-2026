Problem 1:Social Media – Post Engagement Analyzer
1. Problem Statement
Create a function to analyze engagement on a social media post.
Rules
Calculate total likes
If Total Likes ≥ 1000 → Viral Post
Otherwise → Normal Engagement
Real-Time Use
Social media analytics platforms
Marketing dashboards
Hint
Use function
Use loop for calculations
Expected Output Format
Total Likes: 1050
Post Status: Viral Post


```python
def analyze():
    total_likes = 0 # here it will stores the total no.of likes
    n = int(input("How many like counts do you want to enter:")) # here we are asking the user how many like values they want to enter
    for i in range(n): # here loop runs n times to take likes input one by one
        likes = int(input("Enter likes: ")) # Asking the user to enter likes for each time/source
        total_likes += likes # adding the entered likes to total_likes
    if total_likes >= 1000: # Check if total likes are 1000 or more
        status = "Viral Post"
    else:
        status = "Normal Engagement"
    print("Total Likes:", total_likes)
    print("Post Status:", status)
analyze() # calling the function
```

Problem 3: Agriculture – Rainfall Adequacy Checker
1. Problem Statement
Create a function that analyzes seasonal rainfall data.
Rules
Calculate average rainfall
Average ≥ required level → Adequate Rainfall
Otherwise → Inadequate Rainfall
Real-Time Use
Climate monitoring systems
Irrigation planning tools
Hint
Use function
Use loop and math operations
Expected Output Format
Average Rainfall: 72
Rainfall Status: Adequate Rainfall


```python
def rainfall_check():
    total = 0 # here we are storing the total rainfall value
    n = int(input("Enter number of rainfall records: ")) # here we are taking records as input like how many months
    for i in range(n):
        total += int(input("Enter rainfall value: ")) # Taking the rainfall value each time and that has been added to total
    avg = total / n # calculating the average value
    req = int(input("Enter required rainfall level: ")) # taking input from user how much rainfall requried
    print("Average Rainfall:", int(avg))
    print("Rainfall Status:", "Adequate Rainfall" if avg >= req else "Inadequate Rainfall") # checking the condition if average greater than requried
rainfall_check() # calling the function
```

Problem 4: Social Media – Duplicate Account Detection
1. Problem Statement
Create a function to detect duplicate usernames in a system.
Rules
If duplicates exist → Flag accounts
Else → Accounts are unique
Real-Time Use
Platform moderation
Fraud detection systems
Hint
Use function
Use set data structure
Expected Output Format
Duplicate Accounts Found: Yes


```python
def detect_duplicate_accounts():
    n = int(input("Enter number of accounts to check: ")) # asking the user how many user accounts we need to check
    usernames = [] # taking the empty list to store the user names
    for i in range(n):
        usernames.append(input("Enter username: "))
    if len(usernames) != len(set(usernames)): # converting list into set to remove duplicates
        print("Duplicate Accounts Found: Yes")
        print("Action: Flag accounts for review ")
    else:
        print("Duplicate Accounts Found: No")
        print("Action: All accounts are unique ")
detect_duplicate_accounts() # calling the function to run the program
```

Problem 6: Agriculture – Premium Crop Price Filter
1. Problem Statement
Create a function that filters crop prices above a premium threshold.
Rules
Price > ₹2000 → Premium Crop
Else → Regular Crop
Real-Time Use
Market analytics
Export pricing systems
Hint
Use function
Use list filtering logic
Expected Output Format
Premium Crops: [2500, 3200]


```python
def premium_crops():
    prices = [] # taking the empty list for storing
    n = int(input("How many crop prices? "))  # Taking the prices one by one
    for i in range(n):
        prices.append(int(input("Enter price: "))) # Taking the each price and add it to the prices list
    premium = [] # taking an empty list to store the premium crop prices above 2k
    for p in prices: # Checking the each price
        if p > 2000:
            premium.append(p)
    print("Premium Crops:", premium)
premium_crops() # calling the function to run
```

Problem 9: E-Learning – Student Attendance Eligibility System
1. Problem Statement
Create a function to calculate attendance percentage and check exam eligibility.
Rules
Attendance ≥ 75% → Eligible
Attendance < 75% → Not Eligible
Real-Time Use
Learning management systems
Internship tracking platforms
Hint
Use function
Use list and loop
Expected Output Format
Attendance Percentage: 80.0
Exam Eligibility: Eligible


```python
def attendance():
    total = int(input("Total classes: ")) # taking the input as total no.of class
    present = int(input("Classes attended: ")) # taking an input how many classes has been attend by the student
    percent = (present / total) * 100 # calcuating the average attendence of student
    print("Attendance Percentage:", percent)
    print("Exam Eligibility:", "Eligible" if percent >= 75 else "Not Eligible") # checking the conditions
attendance() # calling the function to run the program
```

Problem 10: Smart Electricity Bill Analyzer
1. Problem Statement
You are building a smart electricity billing system for residential users.
Create a function that calculates the electricity bill and classifies usage level based on units consumed.
Billing Rules
First 100 units → ₹3 per unit
Next 100 units (101–200) → ₹5 per unit
Above 200 units → ₹7 per unit
Usage Classification
Total bill < ₹500 → Low Usage
₹500 – ₹1500 → Moderate Usage
₹1500 → High Usage
Real-Time Use
Smart meter systems
Electricity board billing software
Energy consumption analytics
Hint
Use function to calculate bill
Use conditional logic for slabs
Use return values (bill and usage status)


```python
def electricity():
    units = int(input("Enter units consumed: ")) #taking the input from the user how many units consumed
    if units <= 100:
        bill = units*3
    elif units <= 200:
        bill = 100*3+(units-100)*5
    else:
        bill = 100*3+100*5+(units-200)*7
    if bill < 500:
        usage = "Low Usage"
    elif bill <= 1500:
        usage = "Moderate Usage"
    else:
        usage = "High Usage"
    print("Total Bill:", bill)
    print("Usage Level:", usage)
electricity() # calling the function
```
