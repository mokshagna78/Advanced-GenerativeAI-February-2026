Problem 1: Mobile Recharge Validation System
 Problem Statement
Create a function to validate a mobile recharge request.
Rules:
Recharge amount must be ≥ ₹50

Valid plans: 199, 299, 399, 599

User should be allowed to retry if invalid

Real-Time Use
Telecom recharge apps

Payment gateways

Hint
Use function for validation


Use while loop for retry

```python
def validate_recharge():
    valid_plans=[199, 299, 399, 599] #list of validation of plans
    while True: # here we are using infinite to allow retry until correct input
        try:
            amount=int(input("Enter recharge amount: "))
            if amount<50: # checking the amount is less than minimum allowed amount
                print("Minimum recharge is 50rs\n")
                continue # here we are using continue to restart the loop
            plan=int(input("Select plan (199, 299, 399, 599): ")) # asking the user which plan they want to recharge
            if plan not in valid_plans: # checking the plan is in the valid list
                print("Invalid plan selected\n")
                continue # here we are enter continue because the user need to retry again and again
            print("\nRecharge Successful") # if the user given plan and plans in list are matched then we need to print recharge sucessful
            print(f"Amount: {amount}") # it displays user recharge amount
            print(f"Plan: {plan}") # it displays user selected plan
            print("Thank you for using our service")
            break # exit the loop after successful recharge
        except ValueError: # handle the error if the user enters symbols or alphabets or any non-numerical values
            print("Please enter numbers only\n")
validate_recharge() # calling the function to start the recharge process
```

Problem 2: Inventory Reorder Alert System
Problem Statement
Given a dictionary of products and stock quantities:
Print Reorder Alert if stock < 15

Print Stock OK otherwise

Real-Time Use
Warehouse systems

Retail management

Hint
Use function

Use for loop on dictionary

```python
def inventory(products):
    for product,quantity in products.items():
        if quantity<15: # checking stocking in inventory
            print(f"{product}: Reorder Alert (Stock = {quantity})")
        else:
            print(f"{product}: Stock OK (Stock = {quantity})")
ware={} #taking an empty dictionary to store the product in inventory or warehouse
a=int(input("Enter number of products: ")) # asking an input from user how many types of products is there in inventory 
for i in range(a):
    p_name=input(f"Enter product {i+1} name: ") # taking input from user as name of product
    p_qty=int(input(f"Enter {p_name} stock quantity: ")) # taking the input from the product stock
    ware[p_name]=p_qty # storing the values in empty dict
print("\nInventory Status:")
inventory(ware) # calling the function to start progress
```

 Problem 3: Student Result Processing System
Problem Statement
Create a function that:
Accepts a list of student marks

Calculates average marks

Prints Pass if average ≥ 50 else Fail

Real-Time Use
Online exam systems

Evaluation portals

Hint
Use loop to calculate total

Return result from function

```python
def evaluate_student(student_id, name, marks):
    total = sum(marks)  # calculate marksusing the sum function
    average = total / len(marks) # calculating the average

    # checking the pass or fail using below conditions
    if average >= 50:
        result = "PASS"
    else:
        result = "FAIL"
    if average >= 90:
        grade = "A+"
    elif average >= 75:
        grade = "A"
    elif average >= 60:
        grade = "B"
    elif average >= 50:
        grade = "C"
    else:
        grade = "F"
    return {"Student ID":student_id,"Name":name,"Total":total,"Average":average,"Grade":grade,"Result":result}
def exam_portal(): # this is the main system or main function because the validation depends upon this function
    print("===== ONLINE EXAM SYSTEM  PORTAL=====")
    students = []  # here we are taking empty list to store the student reports
    num_students = int(input("Enter number of students: "))
    for s in range(num_students):
        print(f"\nEnter details for Student {s+1}")
        student_id = input("Enter Student ID: ")
        name = input("Enter Student Name: ")
        num_subjects = int(input("Enter number of subjects: "))
        marks = [] # here we are taking an empty list to store student marks
        for i in range(num_subjects): # loop for input marks for each subject
            while True: # loop for validating
                try:
                    mark = int(input(f"Enter marks for Subject {i+1}: "))
                    if 0 <= mark <= 100: # checking valid range of marks
                        marks.append(mark)
                        break # using break for exit validation loop
                    else:
                        print("Marks must be between 0 and 100")
                except ValueError: # handling the non numeric values & symbols
                    print("Enter valid number")
        report = evaluate_student(student_id, name, marks)
        students.append(report)
    print("\n===== EVALUATION REPORT PORTAL=====")
    for student in students:
        print('\n-----------------------------------')
        print(f"Student ID : {student['Student ID']}")
        print(f"Name       : {student['Name']}")
        print(f"Total      : {student['Total']}")
        print(f"Average    : {student['Average']:.2f}")
        print(f"Grade      : {student['Grade']}")
        print(f"Result     : {student['Result']}")
    print("\nAll student evaluations completed.")
exam_portal() #calling the main funtion to run the systems
```

Problem 4: Cab Fare Estimator with Retry Option
Problem Statement
Create a cab fare calculator:
Base fare = ₹50

₹12 per km

Peak hour → 25% extra

Allow retry using while loop

Real-Time Use
Ride-hailing apps

Transport systems

Hint
Function for fare calculation

While loop for retry

```python
def fare_calc(km,peak):
    fare=50+(12*km) # basic fare is 50rs and then per km charge is 12rs        
    if peak=="yes": #checking it is peak hour or not
        fare*=1.25  # if it is peak hour fare is 25% extra            
    return fare
while True: #here we are taking an infinite for getting an retry again and again
    a=float(input("Enter distance (km): "))
    b=input("Peak hour (yes/no): ").lower()
    total=fare_calc(a,b) # calling the function
    print(f"Total Fare: {total:.2f}") #printing total fare amount
    try_again=input("Do You Want Retry (yes/no): ").lower() # here it asking the user if you want to calculate again if user gives no then it exit else it will run again and again
    if try_again!="yes":
        print("Thank you!")
        break
```

Problem 5: 
Employee Attendance Eligibility Checker
Problem Statement
Create a function that:
Accepts attendance list ("P", "A")

Calculates attendance percentage

Returns Eligible if ≥ 75% else Not Eligible

Real-Time Use
HR payroll systems

Internship tracking

 Hint
Use loop to count present days

Use return statement

```python
def attendance(att):
    days_p=att.count("P") # here counting no.of present days as "P"
    percent=(days_p/len(att))*100 # calcuating attendence percentage
    return percent,"Eligible" if percent>=75 else "Not Eligible" # checking the given conditions if greater than 75% they will eligible else not eligible

a=input("Enter attendance (P/A): ").upper().split() # taking an input from user
b,status=attendance(a) # calling the function and storing return values
print(f"Attendance: {b:.2f}%")
print("Status:", status)
```

Problem 6: Password Strength Checker
Problem Statement
Create a function to check password strength.
Rules:
Minimum length = 8


Must contain at least one digit


Must contain at least one special character (@#$)


Real-Time Use
Login systems


Security validation


Hint
Use loop over string


Use conditions inside function

```python
def check(pwd):
    num=spl=False #here we are intilizing variables to check numbers and special charaters
    for p in pwd:
        if p.isdigit(): # checking the charater is a number or not
            num=True
        if p in "@#$": # checking the charater is a special charater or not
            spl=True
    return "Strong" if len(pwd)>=8 and num and spl else "Weak" #checking all conditions we are given above and here
a=input("Enter password: ") # taking an input password from user
print(check(a)) #calling the function
```
