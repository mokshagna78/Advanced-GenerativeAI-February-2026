Problem Statement 1: Employee Performance Bonus Eligibility
Description:
A company evaluates employee performance scores at the end of the year. You are given a dictionary containing employee names and their performance scores.
Requirements:
Identify the highest performance score.
Handle ties if multiple employees have the same highest score.
Display all employees eligible for the top performance bonus.
Input:
employees = {
"Ravi": 92,
"Anita": 88,
"Kiran": 92,
"Suresh": 85
}
Expected Output:
Top Performers Eligible for Bonus: Ravi, Kiran (Score: 92)

```python
employees={
    "Ravi": 92,
    "Anita": 88,
    "Kiran": 92,
    "Suresh": 85
}
high_score=max(employees.values()) #finding highest values through max() and employees.values() gives all the scores
top=[] #Create an empty list to store employees who have the highest score
for name,score in employees.items():
    if score==high_score: #Check if the employee's score equals the highest score
        top.append(name) #If yes add the employee name to the top list
print("Top Performers Eligible for Bonus:", ", ".join(top), f"(Score: {high_score})") #here ", ".join(top) converts list into comma-separated string
```

Problem Statement 2: Search Query Keyword Analysis
Description:
An e-commerce website stores customer search queries. You are given a search query sentence entered by a user.
Requirements:
Convert the input to lowercase.
Ignore common punctuation.
Count the frequency of each keyword.
Display only keywords searched more than once.
Input:
"Buy mobile phone buy phone online"
Expected Output:
{'buy': 2, 'phone': 2}

```python
queries=input().lower()#taking an input from the user and making that in lowercase
for i in '.,!?':
    queries=queries.replace(i,'') #here replacing the punctuation if there is any in user given sentence 
a=queries.split() # here we are using split keyword for making the sentence into list
dic={} # create a empty dict to store the frequency of values
for word in a: # by using this loop counting each words how many times it repeats
    if word in dic:
        dic[word]+=1
    else:
        dic[word]=1
result={} # Create a new dictionary to store only words with frequency values > 1
for word,count in dic.items():
    if count>1:
        result[word]=count
print(result)
```

sensor_readings = [3, 4, 7, 8, 10, 12, 5]
Expected Output:
Valid Sensor Readings (Hour, Value):
[(1, 4), (3, 8), (4, 10), (5, 12)]

Problem Statement 4: Email Domain Usage Analysis
Description:
A company wants to analyze which email providers its users are using. You are given a list of employee email IDs.
Requirements:
Count how many users belong to each email domain.
Calculate the percentage usage of each domain.
Input:
emails = [
"ravi@gmail.com",
"anita@yahoo.com",
"kiran@gmail.com",
"suresh@gmail.com",
"meena@yahoo.com"
]
Expected Output:
gmail.com: 60%
yahoo.com: 40%

```python
a=int(input("enter no.of mail's:")) #taking how many numbers of mails from user
mail=[] #emptylist to store mails
for i in range(a): # taking input mails from user
    m=input(f"Enter email {i+1}: ")
    mail.append(m) # adding mailsto empty list
count={} # empty dict to store counts
for email in mail:
    domain=email.split('@')[1] #get domain after '@'
    if domain in count:
        count[domain]+=1 #increase count if already exist
    else:
        count[domain]=1 # add new domain with count 1
total=len(mail) # no.of mails
print("\nEmail Domain Usage Percentage:")
for domain,d_count in count.items():
    percent=(d_count / total) * 100 #percentage formula
    print(f"{domain}: {percent:.0f}%")
```

Problem Statement 5: Sales Spike Detection
Description:
A retail company tracks daily sales. Sudden spikes in sales may indicate promotions or unusual activity.
Requirements:
Calculate the average daily sales.
Detect days where sales are more than 30% above average.
Display the day number and sale value.
Input:
sales = [1200, 1500, 900, 2200, 1400, 3000]
Expected Output:
Day 4: 2200
Day 6: 3000m

```python
import numpy as np
sales=np.array([1200,1500,900,2200,1400,3000])
avg=np.mean(sales) #Calculating average
ts=avg*1.3 #Calculating threshold (30% above average)
spike_days=np.where(sales>ts)[0] #Boolean masking (Vectorized filtering)
for day in spike_days:
    print(f"Day {day+1}: {sales[day]}") #it cannot print day 4 because 2200 is not greater than 2210
```

Problem Statement 6: Duplicate User ID Detection
Description:
A system stores user IDs during registration. Duplicate IDs can cause data integrity issues.
Requirements:
Identify duplicate user IDs.
Display how many times each duplicate appears.
Input:
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]
Expected Output:
user1 → 3 times
user3 → 2 times

```python
import numpy as np
user_ids = np.array([
    "user1", "user2", "user1",
    "user3", "user1", "user3",
])
unique,count=np.unique(user_ids, return_counts=True) # Detecting duplicates
d=unique[count>1] # finding duplicates
d_count=count[count>1] # count of duplicates
for user,count in zip(d,d_count):
    print(f"{user} → {count} times")
```
