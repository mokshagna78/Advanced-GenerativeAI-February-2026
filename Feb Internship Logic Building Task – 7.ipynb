# Problem 1: Smart Light Controller
1. Problem Statement
Create a class to control a smart light that can be turned ON or OFF and display its status.
🌍 Real-Time Use
Smart home automation systems
📥 Sample Input
Light Name: Bedroom Light
Action: ON
📤 Expected Output
Bedroom Light is ON


```python
class SmartLight:
    def __init__(self, name):
        self.name = name
        self.status = "OFF"
    def turnon(self):
        self.status = "ON"
    def turnoff(self):
        self.status = "OFF"
    def display_status(self):
        print(f"{self.name} Status : {self.status}")
light = SmartLight("Bedroom Light")
light.turnon()
light.display_status()
```

# Problem 2: Employee ID Card System
1. Problem Statement
Create a class that stores employee name, ID, and department and displays ID card details.
Real-Time Use
Corporate HR systems
📥 Sample Input
Employee Name: Rahul
Employee ID: EMP102
Department: IT

📤 Expected Output
Employee ID Card
Name: Rahul
ID: EMP102
Department: IT


```python
class Employee:
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department
    def idcard(self):
        print("Employee ID Card")
        print(f"Name: {self.name}")
        print(f"ID: {self.emp_id}")
        print(f"Department: {self.department}")
employee = Employee("Mokshagna", "2231", "AI&ML Engineer")
employee.idcard()
```

# Problem 3: Mobile Contact Record
1. Problem Statement
Create a class to store contact name and phone number and display contact information.
Real-Time Use
Mobile contact applications
📥 Sample Input
Contact Name: Anita
Phone Number: 9876543210
📤 Expected Output
Contact Saved
Name: Anita
Phone: 9876543210


```python
class Contacts:
    def __init__(self, name, phone_number):
        self.name = name
        self.phone_number = phone_number
    def contact(self):
        print("Contact Saved")
        print(f"Name: {self.name}")
        print(f"Phone: {self.phone_number}")
contact = Contacts("Moksha", "1234567890")
contact.contact()
```

# Problem 4: Product Price Tag Generator
1. Problem Statement
Create a class that stores product name and price and prints a formatted price tag.
 Real-Time Use
Retail billing systems
📥 Sample Input
Product Name: Headphones
Price: 2499
📤 Expected Output
Product: Headphones
Price: ₹2499


```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def tag(self):
        print(f"Product: {self.name}")
        print(f"Price: {self.price}")
product = Product("Headphones",2499)
product.tag()
```

# Problem 5: Movie Rating Display System
1. Problem Statement
Create a class that stores movie name and rating and displays movie details.
Real-Time Use
Streaming platforms
📥 Sample Input
Movie Name: Inception
Rating: 4.8
📤 Expected Output
Movie: Inception
Rating: 4.8 / 5


```python
class Movie:
    def __init__(self, name, rating):
        self.name = name
        self.rating = rating
    def details(self):
        print(f"Movie: {self.name}")
        print(f"Rating: {self.rating}/5")
movie_name = Movie("salaar", 4.6)
movie_name.details()
```

# Problem 6: Delivery Address Manager
1. Problem Statement
Create a class that stores customer name and delivery address and prints delivery details.
Real-Time Use
Courier and logistics platforms
📥 Sample Input
Customer Name: Suman
Address: Hyderabad
📤 Expected Output
Delivery Details
Customer: Suman
Address: Hyderabad


```python
class Delivery:
    def __init__(self, customer_name, address):
        self.customer_name = customer_name
        self.address = address
    def details(self):
        print("Delivery Details")
        print(f"Customer: {self.customer_name}")
        print(f"Address: {self.address}")
delivery = Delivery("Bunny", "KPHB HYD")
delivery.details()
```
